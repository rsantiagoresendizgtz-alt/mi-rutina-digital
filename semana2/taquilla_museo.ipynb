## Descripción
Este programa calcula el precio de entrada al museo según la edad del visitante. Aplica los descuentos correspondientes, evita que se acumulen dos descuentos en un mismo boleto y muestra el corte de caja al cerrar la taquilla.


```python
# Constantes de precios
PRECIO_MENOR = 0.00
PRECIO_NINO = 30.00
PRECIO_ADULTO = 45.00

# Constantes de descuentos
DESCUENTO_ADULTO_MAYOR = 0.12
DESCUENTO_ESTUDIANTE = 0.10
DESCUENTO_PROFESOR = 0.10

# Contador y acumuladores
contador_visitantes = 0
total_descuentos = 0.00
total_a_pagar = 0.00
```

```python
while True:
    entrada = input("Edad del visitante (o 'fin' para cerrar la taquilla): ")

    if entrada.lower() == "fin":
        break

    edad = int(entrada)

    if edad < 0 or edad > 110:
        print("Edad no válida. Intente con el siguiente visitante.")
        continue

    precio_base = 0.00
    descuento = 0.00
    tipo = "No aplica"

    if edad < 3:
        precio_base = PRECIO_MENOR

    elif edad < 18:
        precio_base = PRECIO_NINO

    else:
        precio_base = PRECIO_ADULTO

        if edad >= 60:
            descuento = precio_base * DESCUENTO_ADULTO_MAYOR
            tipo = "Adulto mayor"

        else:
            tipo_opcion = input(
                "Tipo: 1) Estudiante 2) Profesor 3) Ninguno: "
            )

            if tipo_opcion == "1":
                descuento = precio_base * DESCUENTO_ESTUDIANTE
                tipo = "Estudiante"

            elif tipo_opcion == "2":
                descuento = precio_base * DESCUENTO_PROFESOR
                tipo = "Profesor"

            else:
                descuento = 0.00
                tipo = "Ninguno"

    paga = precio_base - descuento

    contador_visitantes += 1
    total_descuentos += descuento
    total_a_pagar += paga

    print(
        f"Visitante {contador_visitantes} | Edad: {edad} | "
        f"{tipo} | Base: ${precio_base:.2f} | "
        f"Descuento: ${descuento:.2f} | A pagar: ${paga:.2f}"
    )

print("\n=== CORTE DE CAJA ===")
print(f"Visitantes cobrados: {contador_visitantes}")
print(f"Total en descuentos: ${total_descuentos:.2f}")
print(f"TOTAL A PAGAR: ${total_a_pagar:.2f}")
```
